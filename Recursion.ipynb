```python
def get_source(g,dest):
    x=g[g['destination']==dest]
    if x.empty:
        return dest
    src=x['source'].iloc[0]
    return get_source(g,src)
```

```python
with open('mapping.json','r') as f:
    tags=json.load(f)
```

```python
%time
rgroup=routes.groupby('Airlines')
source=[]
for i,r in dest.iterrows():
    g=rgroup.get_group(r['Airlines'])
    src=get_source(g,r['destination'])
    #print({'Airlines':r['Airlines'],'Source':src,'Destination':r['destination']})
    source.append({'Airlines':r['Airlines'],'Destination':(r['destination'],tags[r['destination']]),'Source':(src,tags[src])})
```

```python
source=pd.DataFrame(source)
source
```
